```python
from tqix import *
from tqix.pis import *
import numpy as np

numq=50

#create quantum circuit
qc = circuit(numq)
qc.RN(-np.pi/2,0.0)
psi = qc.state

#visualize state
THETA = [0, np.pi]
PHI = [0, 2* np.pi]
husimi_spin_3d(psi.toarray(),THETA , PHI,cmap = 'jet',dirname="./FIG",fname ="husimi_sphere.eps")
```

```python
#get probability
prob = qc.measure(num_shots=1000)
#plot figure
from matplotlib import pyplot as plt
x = np.arange(0,numq+1,1)
plt.bar(x,prob)
plt.savefig("./FIG/Pjm.eps")
```
